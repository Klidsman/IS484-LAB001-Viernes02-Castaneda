# Representación de un problema mediante estados
Parte A: Completar la cha del problema para el siguiente caso: un repartidor debe ir desde el
Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.


|  || 
|-----------|-----------|
| Estado Inicial | Alamacen | 
| Estado Objetivo | Zona Norte |
|Acciones Posibles | Movimientos hasta llegar al objetivos |
|Costo de Cada acción |Mismo costo para cada acción |  


```python
mapa = {
    "Almacen" : ["Centro", "Terminal"],
    "Centro" : ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal" : ["Almacen", "Mercado"],
    "Mercado" : ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

# Implementación de BFS


```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]] # cola de caminos
    visitados = set()

    while cola:
        camino = cola.pop(0) # sacamos el primer camino de la cola
        nodo_actual = camino[-1] #tomamos ultimo elemento de la cola 

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)

    return None # no se encontro un camino

# Ejemplo de uso:
resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

# Vectores y matrices con NumPy

Parte A: Representar cuatro observaciones de un dataset simple (por ejemplo, sensores de
temperatura y humedad) como un vector cada una, y juntarlas en una matriz:

```python
import numpy as np
obs1 = np.array([22.5, 60]) # temperatura, humedad
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])
dataset = np.array([obs1, obs2, obs3, obs4])
print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)
```

# 5.1. Planteamiento del problema


Se quiere encontrar, dentro de un conjunto de observaciones ya conocidas, cuál es la más parecida a
una observación nueva  el mismo principio que usará el algoritmo KNN a partir de la Semana 7.


```python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    """
    pila = [[inicio]]
    visitados = set()

    while pila:
        camino = pila.pop()  # sacar el ÚLTIMO camino de la pila (LIFO)
        nodo_actual = camino[-1]

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                # COMPLETAR: misma logica que en BFS, pero agregando a la pila
                nuevo_camino = camino + [vecino]
                pila.append(nuevo_camino)

    return None

resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
```

# 5.3. Parte B: Encontrar la observación más cercana

```python

obs1 = np.array([22.5, 60])
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])
dataset = np.array([obs1, obs2, obs3, obs4])

nueva_observacion = np.array([20.5, 48])
distancias = []

for i, obs in enumerate(dataset):
    # distancia euclidiana 
    d = np.linalg.norm(nueva_observacion - obs)
    distancias.append(d)
    print(f"Distancia a la observacion {i}: {d:.2f}")

indice_mas_cercano = np.argmin(distancias)
print("La observacion mas parecida es la numero", indice_mas_cercano)
```

# 5.4. Parte C: Reflexíon

## por qué la distancia euclidiana es una forma razonable de medir qué tan parecidas son dos observaciones?

Porque si dos puntos están numéricamente cerca en todas sus dimensiones, representan estados del entorno casi idénticos.

## En qué casos podría no ser la mejor opción (por ejemplo, si las características tienen escalas muy distintas, como soles vs. años)?

En escalas o unidades distintas, alta dimensionalidad con muchas columnas, no se pueden calcular datos categoricos como sexo, tipo cliente, etc
