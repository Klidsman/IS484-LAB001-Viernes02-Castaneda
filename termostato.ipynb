```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22.0, margen=1.0):
    """
    Simula el proceso de toma de decisiones de un agente reactivo simple para control térmico.

    Parámetros:
        temperatura_actual (float): Lectura actual capturada por el sensor (°C).
        temperatura_objetivo (float): Temperatura deseada por el usuario (por defecto: 22.0°C).
        margen (float): Margen de tolerancia o histéresis térmica (por defecto: 1.0°C).

    Retorna:
        dict: Estado booleano de los actuadores y descripción operativa del agente.
    """
    # Cálculo del error térmico (percepción procesada)
    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen:
        # Condición: La temperatura supera el umbral superior (hace calor)
        aire_acondicionado = True
        calefaccion = False
        estado = "ENFRIANDO (Aire Acondicionado encendido)"

    elif diferencia < -margen:
        # Condición: La temperatura cae por debajo del umbral inferior (hace frío)
        aire_acondicionado = False
        calefaccion = True
        estado = "CALENTANDO (Calefacción encendida)"

    else:
        # Condición: La temperatura se encuentra en el rango de confort
        aire_acondicionado = False
        calefaccion = False
        estado = "INACTIVO (Temperatura en rango óptimo)"

    return {
        "temperatura_medida": temperatura_actual,
        "diferencia": round(diferencia, 2),
        "aire_acondicionado": aire_acondicionado,
        "calefaccion": calefaccion,
        "estado": estado
    }


# Casos de prueba
lecturas_prueba = [25.5, 18.0, 22.4, 20.8]

for temp in lecturas_prueba:
    resultado = agente_termostato(temperatura_actual=temp)
    print(f"Sensor: {resultado['temperatura_medida']}°C | "
          f"Dif: {resultado['diferencia']:+.2f}°C -> {resultado['estado']}")
```

# Laboratorio: Agente Inteligente para Control de Temperatura

## 1. Especificación del Entorno de Tareas (PEAS)
Para modelar el agente termostato bajo el paradigma de agentes inteligentes, se define su estructura PEAS:

* **P ( Rendimiento):** Mantener la temperatura dentro del rango de confort deseado ($T_{\text{objetivo}} \pm \text{margen}$), minimizar el consumo de energía eléctrica y evitar el desgaste mecánico de los actuadores por conmutaciones frecuentes.
* **E ( Entorno):** Recinto cerrado (vivienda, habitación u oficina) sujeto a perturbaciones térmicas internas (presencia de personas) y externas (clima).
* **A ( Actuadores):** Compresor del aire acondicionado (enfriamiento), resistencia o caldera (calefacción), y relé de corte general.
* **S ( Sensores):** Termistor o sensor de temperatura digital que captura la lectura del ambiente (`temperatura_actual`).

## 2. Percepción y Lógica de Decisión (Histéresis)
El agente toma como entrada la percepción del sensor y calcula la desviación respecto a la meta:

$$\text{diferencia} = T_{\text{actual}} - T_{\text{objetivo}}$$

Para evitar que el sistema encienda y apague repetidamente ante fluctuaciones mínimas de decimales, se introduce un parámetro de tolerancia llamado **margen** (o banda muerta):
* Si la diferencia excede el margen positivo, el ambiente está sobrecalentado.
* Si la diferencia desciende por debajo del margen negativo, el ambiente está frío.
* Si la diferencia está contenida en $[-\text{margen}, +\text{margen}]$, el sistema no requiere intervención.

```python
def calcular_error_termico(temperatura_actual, temperatura_objetivo=22.0):
    """Calcula la discrepancia entre la percepción actual y la meta del agente."""
    return temperatura_actual - temperatura_objetivo
```

## 3. Función del Agente Termostato
Implementamos las reglas condición-acción (agente reactivo simple basado en umbrales). El agente devuelve el estado booleano de cada actuador y un reporte textual de su estado de operación.

```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22.0, margen=1.0):
    """
    Agente reactivo simple para control térmico con margen de tolerancia.
    """
    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen:
        aire_acondicionado = True
        calefaccion = False
        estado = "ENFRIANDO (Aire Acondicionado encendido)"

    elif diferencia < -margen:
        aire_acondicionado = False
        calefaccion = True
        estado = "CALENTANDO (Calefacción encendida)"

    else:
        aire_acondicionado = False
        calefaccion = False
        estado = "INACTIVO (Zona de confort alcanzada)"

    return {
        "temperatura_medida": temperatura_actual,
        "diferencia": round(diferencia, 2),
        "aire_acondicionado": aire_acondicionado,
        "calefaccion": calefaccion,
        "estado": estado
    }
```
