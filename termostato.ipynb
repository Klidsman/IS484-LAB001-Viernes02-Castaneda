```python
def agente_termostato(temperatura_actual, temperatura_objetivo=22.0, margen=1.0):
    """
    Simula el proceso de toma de decisiones de un agente reactivo simple para control térmico.

    Parámetros:
        temperatura_actual (float): Lectura actual capturada por el sensor (°C).
        temperatura_objetivo (float): Temperatura deseada por el usuario (por defecto: 22.0°C).
        margen (float): Margen de tolerancia o histéresis térmica (por defecto: 1.0°C).

    Retorna:
        dict: Estado booleano de los actuadores y descripción operativa del agente.
    """
    # Cálculo del error térmico (percepción procesada)
    diferencia = temperatura_actual - temperatura_objetivo

    if diferencia > margen:
        # Condición: La temperatura supera el umbral superior (hace calor)
        aire_acondicionado = True
        calefaccion = False
        estado = "ENFRIANDO (Aire Acondicionado encendido)"

    elif diferencia < -margen:
        # Condición: La temperatura cae por debajo del umbral inferior (hace frío)
        aire_acondicionado = False
        calefaccion = True
        estado = "CALENTANDO (Calefacción encendida)"

    else:
        # Condición: La temperatura se encuentra en el rango de confort
        aire_acondicionado = False
        calefaccion = False
        estado = "INACTIVO (Temperatura en rango óptimo)"

    return {
        "temperatura_medida": temperatura_actual,
        "diferencia": round(diferencia, 2),
        "aire_acondicionado": aire_acondicionado,
        "calefaccion": calefaccion,
        "estado": estado
    }


# Casos de prueba
lecturas_prueba = [25.5, 18.0, 22.4, 20.8]

for temp in lecturas_prueba:
    resultado = agente_termostato(temperatura_actual=temp)
    print(f"Sensor: {resultado['temperatura_medida']}°C | "
          f"Dif: {resultado['diferencia']:+.2f}°C -> {resultado['estado']}")
```
